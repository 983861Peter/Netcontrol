```python
import requests

API_URL = "http://localhost:8080/routers"

try:
    response = requests.get(API_URL)
    print("STATUS CODE:", response.status_code)

    if response.status_code == 200:
        print("✅ Endpoint reachable!")
        print("📌 Response JSON:")
        print(response.json())
    else:
        print("❌ Endpoint returned error:")
        print(response.text)

except Exception as e:
    print("❌ Network/Connection error:", str(e))

```
